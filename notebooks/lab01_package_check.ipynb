# AI Project Lab 01 – Foundations of AI Engineering

## 1. Problem Statement
Building Artificial Intelligence projects requires a standardized project structure, an isolated virtual environment, and version control. Without proper setup, projects become disorganized, dependencies conflict with each other, and progress cannot be tracked properly.

## 2. Core Objectives
- Create a **standardized AI project folder structure** suitable for production workflows.
- Configure an **isolated Python virtual environment (`aipdd_env`)** using `venv`.
- Install and pin core AI libraries (**NumPy, OpenCV, PyTorch, Matplotlib, Ultralytics, and Jupyter**) in `requirements.txt`.
- Initialize **Git version control** and apply a feature-branch workflow (`feature/lab01-setup`).

### Environment Setup Commands
```bash
py -3.11 -m venv aipdd_env
aipdd_env\Scripts\activate
pip install -r requirements.txt
```
